fig_sigma — Fig. `fig:sigma` (impurity self-energy, strongly damped model)
-Im and Re of Sigma^R_up(omega) at the steady state from the empty impurity, gamma = 4, 8, 16.
Reads `data/` (default) or a fresh `output/` (`USE_OUTPUT = True`) and writes `fig_sigma.pdf`.

```python
USE_OUTPUT = False
OUTPUT_DIR = "output"
```

```python
import sys
from pathlib import Path
import numpy as np
HERE = Path.cwd()
sys.path.insert(0, str(HERE.parent))
from common import io as IO
from common.style import plt, W2, style, panel, tag, legend_handles, LABEL, save
if USE_OUTPUT:
    import postprocess
    D = postprocess.build(OUTPUT_DIR)
else:
    D = IO.read_data(HERE / "data")
S = D["sigma.h5"]
METHODS = ["HF", "TPPX", "TPPX_0", "DSL"]
GAMMAS = [4.0, 8.0, 16.0]
Z = {"exact": 2, "HF": 3, "TPPX": 3, "TPPX_0": 3, "DSL": 4}
sig = lambda m, g: S[m][f"c3_ic00_g{g:g}"]
```

```python
fig, ax = plt.subplots(2, 3, figsize=(W2, 4.0), sharex=True)
fig.subplots_adjust(left=0.085, right=0.99, top=0.87, bottom=0.10, hspace=0.18, wspace=0.27)
for j, g in enumerate(GAMMAS):
    for i, q in enumerate((lambda s: -s.imag, lambda s: s.real)):
        a = ax[i, j]
        for m in ["exact"] + METHODS:
            r = sig(m, g); a.plot(r["omega"], q(r["Sigma"][:, 0]), **style(m), zorder=Z[m])
        a.set_xlim(-6, 6); a.set_xticks([-6, -3, 0, 3, 6])
        if i == 0:
            tag(a, rf"$\gamma={g:g}$", "ur")
        if j == 0:
            a.set_ylabel(r"$-\mathrm{Im}\,\Sigma^R(\omega)$" if i == 0 else r"$\mathrm{Re}\,\Sigma^R(\omega)$")
    ax[1, j].set_xlabel(r"$\omega$")
for k, a in enumerate(ax.ravel()):
    panel(a, "(%s)" % "abcdef"[k])
M = ["exact"] + METHODS
fig.legend(legend_handles(M), [LABEL[m] for m in M], loc="upper center", ncol=5, bbox_to_anchor=(0.53, 1.0))
fig.align_ylabels(ax[:, 0])
save(fig, HERE / "fig_sigma")
```
